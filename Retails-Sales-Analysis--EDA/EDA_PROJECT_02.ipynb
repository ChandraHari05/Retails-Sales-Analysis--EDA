```python
plt.figure(figsize=(12, 6)) 
sns.scatterplot(y='Age',x='Customer ID',data=df)
```

```python
sns.boxplot(x="Total Amount",data=df)
```

```python
sns.histplot(x="Total Amount",data=df)
```

```python
d1 = df.groupby("year")["Quantity"].mean()
d1.plot(kind="line")
plt.xlabel("Year")
plt.ylabel("Average Quantity")
plt.title("Average Quantity by Year")
plt.show()
```

```python
d2 = df.groupby("month")["Quantity"].mean()
d2.plot(kind="line")
plt.xlabel("month")
plt.ylabel("Average Quantity")
plt.title("Average Quantity by Month")
plt.show()
```

```python
top = df.sort_values(by='Quantity', ascending=False)
plt.figure(figsize=(18, 5))
sns.barplot(data=top, x='Product Category', y='Quantity', hue='Product Category', palette='viridis', dodge=False) 
plt.title('product category vs total amount')
plt.xlabel('Product Category')
plt.ylabel('Age')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
df.groupby('Product Category')['Age'].count().plot.pie(
    autopct='%1.1f%%',
    figsize=(8, 8),
    title='Product Category by Age'
)
plt.ylabel('')
plt.show()
```
